## Interpretation notes

- Predictive discrimination and probability calibration answer different questions.
- SHAP summaries describe the fitted prediction function, not causal effects.
- Marginal conformal coverage does not guarantee comparable coverage within each outcome class.
- Class-conditional calibration can improve minority-outcome coverage while increasing ambiguous prediction sets.
